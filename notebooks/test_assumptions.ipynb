```python
import numpy as np
nomic = sentence_transformers.SentenceTransformer(
            "nomic-ai/nomic-embed-text-v1.5",
            trust_remote_code=True,
            device="cpu",
        )

v1 = nomic.encode(sentence_1, normalize_embeddings=True)
v2 = nomic.encode(sentence_2, normalize_embeddings=True)
v3 = nomic.encode(sentence_3, normalize_embeddings=True)
v4 = nomic.encode(sentence_4, normalize_embeddings=True)
print(np.linalg.norm(v1 - v2))
print(np.linalg.norm(v2 - v3))
print(np.linalg.norm(v1 - v3))
print(np.linalg.norm(v1-v4))
```
